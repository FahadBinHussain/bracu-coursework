```python
#INPUTS
graph = {'A': {'B': 75, 'C': 118,'E': 140}, 'B': {'A': 75}, 'C': {'A': 118,'D': 111}, 'D': {'C': 111},'E': {'A': 140, 'G': 80, 'F': 99}, 'F': {'E': 99, 'I': 211}, 'G': {'E':80, 'H': 97}, 'H': {'G': 97, 'I': 101}}
heuristics = {'A': 366, 'B': 374, 'C': 329, 'D': 244, 'E': 253, 'F': 178, 'G': 193, 'H': 98, 'I': 0}
```

```python
import heapq
def A_star_search(graph, heuristics, start, goal):
  queue = [] # Use it as priority Queue
  visited = [] # Use it to track the visited nodes

  queue.append((heuristics[start], 0, [start], start))

  while queue:
    best_path_index = 0
    for i in range(1, len(queue)):
      if queue[i][0] < queue[best_path_index][0]:
        best_path_index = i
    f_cost, g_cost, path, current_node = queue.pop(best_path_index)
    if current_node in visited:
      continue
    visited.append(current_node)
    if current_node == goal:
      return path, g_cost

    for neighbor, cost in graph.get(current_node, {}).items():
      if neighbor not in visited:
        new_g_cost = g_cost + cost
        new_f_cost = new_g_cost + heuristics[neighbor]
        queue.append((new_f_cost, new_g_cost, path + [neighbor], neighbor))

  return None, None  # Path, Cost
```

```python
path, cost = A_star_search(graph, heuristics, 'A', 'I')
print("Path =", path)   # Path = ['A', 'E', 'G', 'H', 'I']
print("Cost =", cost)   # Cost = 418
```
